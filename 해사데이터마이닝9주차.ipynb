```python
import pandas as pd
from scipy import stats
import numpy as np

df = pd.read_csv("/content/drive/MyDrive/type3_sample_csvs/delivery_time.csv")
x = df["delivery_min"].dropna()


result = stats.ttest_1samp(x, popmean=30, alternative="less")

mean_x = x.mean()
n = len(x)
se = stats.sem(x)
ci_low, ci_high = stats.t.interval(
0.95, df=n-1, loc=mean_x, scale=se
)

print("표본평균:", mean_x)
print("t-statistic:", result.statistic)
print("p-value:", result.pvalue)
print("95% CI:", (ci_low, ci_high))
```

```python
import pandas as pd
from scipy import stats
df = pd.read_csv("/content/drive/MyDrive/type3_sample_csvs/before_after_worktime.csv")
before = df["before_min"]
after = df["after_min"]
diff = before - after
result = stats.ttest_rel(before, after, alternative="greater")
print("교육 전 평균:", before.mean())
print("교육 후 평균:", after.mean())
print("평균 차이(before - after):", diff.mean())
print("t-statistic:", result.statistic)
print("p-value:", result.pvalue)
```
